````python
# ============================================================
# Prerequisites
# uv add requests pandas
# ============================================================
import requests
import json
import pandas as pd
from IPython.display import display

POLARIS_URL = 'http://192.168.139.2:8181'
REALM       = 'POLARIS'
BASE        = f'{POLARIS_URL}/api/management/v1'
CAT         = f'{POLARIS_URL}/api/catalog/v1'

def get_token(client_id='root', client_secret='polaris-secret', realm=REALM):
    r = requests.post(
        f'{POLARIS_URL}/api/catalog/v1/oauth/tokens',
        headers={'Polaris-Realm': realm},
        data={'grant_type': 'client_credentials', 'client_id': client_id,
              'client_secret': client_secret, 'scope': 'PRINCIPAL_ROLE:ALL'}
    )
    return r

def h(token, realm=REALM):
    return {'Authorization': f'Bearer {token}', 'Polaris-Realm': realm,
            'Content-Type': 'application/json'}

# Test result tracker
RESULTS = []

def test(group, name, r, expected_status, note=''):
    actual = r.status_code
    passed = actual == expected_status
    icon   = '✅' if passed else '❌'
    result = {
        'group': group, 'test': name,
        'expected': expected_status, 'actual': actual,
        'passed': passed, 'note': note
    }
    RESULTS.append(result)
    err = ''
    if r.text:
        try:
            err = r.json().get('error', {}).get('message', '')[:80]
        except:
            err = r.text[:80]
    print(f'  {icon} [{actual}] {name}')
    if not passed or note:
        if err: print(f'       → {err}')
        if note: print(f'       💡 {note}')
    return r

TOKEN_ROOT = get_token().json()['access_token']
print('✅ Setup complete')
````

## Group 1 — Authentication Edge Cases

````python
print('=== Group 1: Authentication Edge Cases ===')

# 1. Wrong realm header
r = requests.post(f'{POLARIS_URL}/api/catalog/v1/oauth/tokens',
    headers={'Polaris-Realm': 'WRONG-REALM'},
    data={'grant_type': 'client_credentials', 'client_id': 'root',
          'client_secret': 'polaris-secret', 'scope': 'PRINCIPAL_ROLE:ALL'})
test('Auth', 'Wrong realm header', r, 404, 'Realm not found')

# 2. Wrong client_secret
r = requests.post(f'{POLARIS_URL}/api/catalog/v1/oauth/tokens',
    headers={'Polaris-Realm': REALM},
    data={'grant_type': 'client_credentials', 'client_id': 'root',
          'client_secret': 'WRONG-SECRET', 'scope': 'PRINCIPAL_ROLE:ALL'})
test('Auth', 'Wrong client_secret', r, 401)

# 3. Wrong client_id
r = requests.post(f'{POLARIS_URL}/api/catalog/v1/oauth/tokens',
    headers={'Polaris-Realm': REALM},
    data={'grant_type': 'client_credentials', 'client_id': 'nonexistent',
          'client_secret': 'polaris-secret', 'scope': 'PRINCIPAL_ROLE:ALL'})
test('Auth', 'Wrong client_id', r, 401)

# 4. No realm header
r = requests.post(f'{POLARIS_URL}/api/catalog/v1/oauth/tokens',
    headers={'Content-Type': 'application/x-www-form-urlencoded'},
    data={'grant_type': 'client_credentials', 'client_id': 'root',
          'client_secret': 'polaris-secret', 'scope': 'PRINCIPAL_ROLE:ALL'})
test('Auth', 'No realm header', r, 404, 'requireHeader=true in config')

# 5. Correct auth
r = get_token()
test('Auth', 'Correct credentials', r, 200)
TOKEN_ROOT = r.json()['access_token']

# 6. Root token on Catalog API (namespace listing)
r = requests.get(f'{CAT}/test-catalog/namespaces', headers=h(TOKEN_ROOT))
test('Auth', 'Root token on Catalog API', r, 404, 'Root has no catalog role')

# 7. Invalid token format
r = requests.get(f'{BASE}/principals',
    headers={'Authorization': 'Bearer INVALID-TOKEN', 'Polaris-Realm': REALM})
test('Auth', 'Invalid token format', r, 401)

# 8. Missing Authorization header
r = requests.get(f'{BASE}/principals', headers={'Polaris-Realm': REALM})
test('Auth', 'Missing Authorization header', r, 401)
````

## Group 2 — Principal Role Dependencies

````python
# Verify root has correct permissions first
r = requests.get(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT))
if r.status_code == 403:
    print('❌ Root missing SERVICE_MANAGE_ACCESS — fix grants first')
    print('   Run: INSERT INTO grant_records...')
else:
    print(f'✅ Root verified — found {len(r.json().get("roles", []))} principal roles')
    print('   Ready to run tests')
````

````python
print('=== Group 2: Principal Role Dependencies ===')

# Setup
requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'test-pr'}})
requests.post(f'{BASE}/principals', headers=h(TOKEN_ROOT),
    json={'name': 'test-principal-pr', 'type': 'SERVICE'})
requests.put(f'{BASE}/principals/test-principal-pr/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'test-pr'}})

# 1. Create principal role
r = requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'dep-test-pr'}})
test('PrincipalRole', 'Create principal role', r, 201)

# 2. Create duplicate principal role
r = requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'dep-test-pr'}})
test('PrincipalRole', 'Create duplicate principal role', r, 409)

# 3. Delete principal role WITH principal assigned
r = requests.delete(f'{BASE}/principal-roles/test-pr', headers=h(TOKEN_ROOT))
test('PrincipalRole', 'Delete role with principal assigned', r, 204,
     'Cascade or block? Check if principal still exists')

# Verify principal still exists after role deletion
r = requests.get(f'{BASE}/principals/test-principal-pr', headers=h(TOKEN_ROOT))
test('PrincipalRole', 'Principal exists after role deleted', r, 200,
     'Principal survives — role assignment cascaded')

# 4. Get deleted principal role
r = requests.get(f'{BASE}/principal-roles/test-pr', headers=h(TOKEN_ROOT))
test('PrincipalRole', 'Get deleted principal role', r, 404)

# 5. Delete non-existent principal role
r = requests.delete(f'{BASE}/principal-roles/nonexistent-role', headers=h(TOKEN_ROOT))
test('PrincipalRole', 'Delete non-existent principal role', r, 404)

# Cleanup
requests.delete(f'{BASE}/principal-roles/dep-test-pr', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/principals/test-principal-pr', headers=h(TOKEN_ROOT))
````

## Group 3 — Principal Dependencies

````python
print('=== Group 3: Principal Dependencies ===')

# Setup
requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'test-pr-for-principal'}})

# 1. Create principal
r = requests.post(f'{BASE}/principals', headers=h(TOKEN_ROOT),
    json={'name': 'dep-test-user', 'type': 'SERVICE'})
test('Principal', 'Create principal', r, 201)
creds = r.json().get('credentials', {})
TEST_CID = creds.get('clientId', '')
TEST_CSECRET = creds.get('clientSecret', '')

# 2. Create duplicate principal
r = requests.post(f'{BASE}/principals', headers=h(TOKEN_ROOT),
    json={'name': 'dep-test-user', 'type': 'SERVICE'})
test('Principal', 'Create duplicate principal', r, 409)

# 3. Assign non-existent principal role
r = requests.put(f'{BASE}/principals/dep-test-user/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'nonexistent-role'}})
test('Principal', 'Assign non-existent principal role', r, 404)

# 4. Assign valid principal role
r = requests.put(f'{BASE}/principals/dep-test-user/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'test-pr-for-principal'}})
test('Principal', 'Assign valid principal role', r, 201)

# 5. Assign same principal role again
r = requests.put(f'{BASE}/principals/dep-test-user/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'test-pr-for-principal'}})
test('Principal', 'Assign same principal role again', r, 500,
     'Idempotent or error?')

# 6. Self-rotate credentials
if TEST_CID and TEST_CSECRET:
    user_token = get_token(TEST_CID, TEST_CSECRET).json().get('access_token', '')
    if user_token:
        r = requests.post(f'{BASE}/principals/dep-test-user/rotate',
            headers=h(user_token))
        test('Principal', 'Self-rotate credentials', r, 200)

# 7. Root rotate other principal
r = requests.post(f'{BASE}/principals/dep-test-user/rotate', headers=h(TOKEN_ROOT))
test('Principal', 'Root rotate other principal', r, 403,
     'Admin cannot rotate others — by design')

# 8. Delete principal with role assigned
r = requests.delete(f'{BASE}/principals/dep-test-user', headers=h(TOKEN_ROOT))
test('Principal', 'Delete principal with role assigned', r, 204,
     'Cascade or block?')

# 9. Get deleted principal
r = requests.get(f'{BASE}/principals/dep-test-user', headers=h(TOKEN_ROOT))
test('Principal', 'Get deleted principal', r, 404)

# Cleanup
requests.delete(f'{BASE}/principal-roles/test-pr-for-principal', headers=h(TOKEN_ROOT))
````

## Group 4 — Catalog Dependencies

````python
print('=== Group 4: Catalog Dependencies ===')

def make_catalog(name, location):
    return {
        'catalog': {
            'name': name, 'type': 'INTERNAL',
            'properties': {'default-base-location': location},
            'storageConfigInfo': {
                'storageType': 'S3',
                'allowedLocations': ['s3a://data-catalog-bucket/'],
                'roleArn': 'arn:aws:iam::123456789012:role/dummy',
                'pathStyleAccess': True
            }
        }
    }

# Add at top of Group 4 setup
requests.delete(f'{BASE}/catalogs/dep-test-catalog', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/catalogs/dep-test-catalog-s3', headers=h(TOKEN_ROOT))
import time; time.sleep(1)

# 1. Create catalog with s3a://
r = requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json=make_catalog('dep-test-catalog', 's3a://data-catalog-bucket/dep-test/'))
test('Catalog', 'Create catalog with s3a://', r, 201)

# 2. Create duplicate catalog
r = requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json=make_catalog('dep-test-catalog', 's3a://data-catalog-bucket/dep-test/'))
test('Catalog', 'Create duplicate catalog', r, 409)

# 3. Create catalog with s3:// (not s3a://)
r = requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json=make_catalog('dep-test-catalog-s3', 's3://data-catalog-bucket/dep-test-s3/'))
test('Catalog', 'Create catalog with s3:// scheme', r, 201,
     '✅ FINDING: Polaris accepts s3:// at creation — no scheme validation')
if r.status_code == 201:
    requests.delete(f'{BASE}/catalogs/dep-test-catalog-s3', headers=h(TOKEN_ROOT))

# 4. Create namespace inside catalog
r = requests.post(f'{CAT}/dep-test-catalog/namespaces', headers=h(TOKEN_ROOT),
    json={'namespace': ['dep-ns'], 'properties': {}})
test('Catalog', 'Create namespace (via root — should fail)', r, 500,
     'Root cannot use Catalog API')

# 5. Delete catalog with catalog_admin role inside
r = requests.delete(f'{BASE}/catalogs/dep-test-catalog', headers=h(TOKEN_ROOT))
test('Catalog', 'Delete catalog (has catalog_admin role)', r, 400,
     'Does Polaris block or cascade delete of catalog roles?')

# 6. Get deleted catalog
r = requests.get(f'{BASE}/catalogs/dep-test-catalog', headers=h(TOKEN_ROOT))
test('Catalog', 'Get deleted catalog', r, 404)

# 7. Delete non-existent catalog
r = requests.delete(f'{BASE}/catalogs/nonexistent-catalog', headers=h(TOKEN_ROOT))
test('Catalog', 'Delete non-existent catalog', r, 404)
````

## Group 5 — Catalog Role Dependencies

````python
print('=== Group 5: Catalog Role Dependencies ===')

# Setup
requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json={'catalog': {'name': 'cr-test-catalog', 'type': 'INTERNAL',
        'properties': {'default-base-location': 's3a://data-catalog-bucket/cr-test/'},
        'storageConfigInfo': {'storageType': 'S3',
            'allowedLocations': ['s3a://data-catalog-bucket/'],
            'roleArn': 'arn:aws:iam::123456789012:role/dummy',
            'pathStyleAccess': True}}})
requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'cr-test-pr'}})

# 1. Create catalog role
r = requests.post(f'{BASE}/catalogs/cr-test-catalog/catalog-roles', headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'cr-test-role'}})
test('CatalogRole', 'Create catalog role', r, 201)

# 2. Create duplicate catalog role
r = requests.post(f'{BASE}/catalogs/cr-test-catalog/catalog-roles', headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'cr-test-role'}})
test('CatalogRole', 'Create duplicate catalog role', r, 409)

# 3. Grant privilege
r = requests.put(
    f'{BASE}/catalogs/cr-test-catalog/catalog-roles/cr-test-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'catalog', 'privilege': 'CATALOG_READ_PROPERTIES'}})
test('CatalogRole', 'Grant privilege to catalog role', r, 201)

# 4. Grant duplicate privilege
r = requests.put(
    f'{BASE}/catalogs/cr-test-catalog/catalog-roles/cr-test-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'catalog', 'privilege': 'CATALOG_READ_PROPERTIES'}})
test('CatalogRole', 'Grant duplicate privilege', r, 500,
     'Duplicate key in grant_records — known 500 behavior')

# 5. Grant to non-existent catalog role
r = requests.put(
    f'{BASE}/catalogs/cr-test-catalog/catalog-roles/nonexistent-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'catalog', 'privilege': 'CATALOG_READ_PROPERTIES'}})
test('CatalogRole', 'Grant to non-existent catalog role', r, 404)

# Delete namespace first, then catalog
requests.delete(f'{CAT}/dep-test-catalog/namespaces/dep-ns', headers=h(TOKEN_ROOT))
r = requests.delete(f'{BASE}/catalogs/dep-test-catalog', headers=h(TOKEN_ROOT))
test('Catalog', 'Delete empty catalog', r, 204)

# 6. Assign catalog role to principal role
r = requests.put(
    f'{BASE}/principal-roles/cr-test-pr/catalog-roles/cr-test-catalog',
    headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'cr-test-role'}})
test('CatalogRole', 'Assign catalog role to principal role', r, 201)

# 7. Delete catalog role with grants AND principal role assignment
r = requests.delete(
    f'{BASE}/catalogs/cr-test-catalog/catalog-roles/cr-test-role',
    headers=h(TOKEN_ROOT))
test('CatalogRole', 'Delete catalog role with grants + assignments', r, 204,
     'Cascade: grants and PR assignments deleted too?')

# 8. Verify principal role no longer has the catalog role
r = requests.get(
    f'{BASE}/principal-roles/cr-test-pr/catalog-roles/cr-test-catalog',
    headers=h(TOKEN_ROOT))
roles = r.json().get('roles', [])
print(f'  💡 Principal role catalog roles after deletion: {roles}')

# 9. Create role on non-existent catalog
r = requests.post(f'{BASE}/catalogs/nonexistent-catalog/catalog-roles',
    headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'test'}})
test('CatalogRole', 'Create role on non-existent catalog', r, 404)

# Cleanup
requests.delete(f'{BASE}/catalogs/cr-test-catalog', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/principal-roles/cr-test-pr', headers=h(TOKEN_ROOT))
````

## Group 6 — Namespace Dependencies

````python
print('=== Group 6: Namespace Dependencies ===')

# Setup — need service token for Catalog API
# Create a test catalog and service principal with access
requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json={'catalog': {'name': 'ns-test-catalog', 'type': 'INTERNAL',
        'properties': {'default-base-location': 's3a://data-catalog-bucket/ns-test/'},
        'storageConfigInfo': {'storageType': 'S3',
            'allowedLocations': ['s3a://data-catalog-bucket/'],
            'roleArn': 'arn:aws:iam::123456789012:role/dummy',
            'pathStyleAccess': True}}})
requests.post(f'{BASE}/catalogs/ns-test-catalog/catalog-roles', headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'ns-admin-role'}})
requests.put(f'{BASE}/catalogs/ns-test-catalog/catalog-roles/ns-admin-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'catalog', 'privilege': 'CATALOG_MANAGE_CONTENT'}})
requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'ns-test-pr'}})
requests.put(f'{BASE}/principal-roles/ns-test-pr/catalog-roles/ns-test-catalog',
    headers=h(TOKEN_ROOT), json={'catalogRole': {'name': 'ns-admin-role'}})
r = requests.post(f'{BASE}/principals', headers=h(TOKEN_ROOT),
    json={'name': 'ns-test-svc', 'type': 'SERVICE'})
creds = r.json().get('credentials', {})
requests.put(f'{BASE}/principals/ns-test-svc/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'ns-test-pr'}})
SVC_TOKEN = get_token(creds.get('clientId', ''), creds.get('clientSecret', '')).json().get('access_token', '')
print(f'  Service token: {"✅" if SVC_TOKEN else "❌"}')

# 1. Create top-level namespace
r = requests.post(f'{CAT}/ns-test-catalog/namespaces', headers=h(SVC_TOKEN),
    json={'namespace': ['parent'], 'properties': {}})
test('Namespace', 'Create top-level namespace', r, 200)

# 2. Create nested namespace
r = requests.post(f'{CAT}/ns-test-catalog/namespaces', headers=h(SVC_TOKEN),
    json={'namespace': ['parent', 'child'], 'properties': {}})
test('Namespace', 'Create nested namespace (parent.child)', r, 200)

# 3. Create deeply nested namespace
r = requests.post(f'{CAT}/ns-test-catalog/namespaces', headers=h(SVC_TOKEN),
    json={'namespace': ['parent', 'child', 'grandchild'], 'properties': {}})
test('Namespace', 'Create deeply nested namespace (3 levels)', r, 200)

# 4. Create duplicate namespace
r = requests.post(f'{CAT}/ns-test-catalog/namespaces', headers=h(SVC_TOKEN),
    json={'namespace': ['parent'], 'properties': {}})
test('Namespace', 'Create duplicate namespace', r, 409)

# 5. Create child without parent existing
r = requests.post(f'{CAT}/ns-test-catalog/namespaces', headers=h(SVC_TOKEN),
    json={'namespace': ['nonexistent-parent', 'child'], 'properties': {}})
test('Namespace', 'Create child without parent existing', r, 404,
     'Does Polaris auto-create parent or reject?')

# 6. Create table inside namespace
r = requests.post(f'{CAT}/ns-test-catalog/namespaces/parent/tables', headers=h(SVC_TOKEN),
    json={'name': 'dep-test-table',
          'schema': {'type': 'struct', 'schema-id': 0,
              'fields': [{'id': 1, 'name': 'id', 'type': 'long', 'required': True}]},
          'partition-spec': {'spec-id': 0, 'fields': []},
          'write-order': {'order-id': 0, 'fields': []},
          'stage-create': False, 'properties': {}})
test('Namespace', 'Create table inside namespace', r, 200)

# 7. Delete namespace WITH table inside
r = requests.delete(f'{CAT}/ns-test-catalog/namespaces/parent', headers=h(SVC_TOKEN))
test('Namespace', 'Delete namespace with table inside', r, 400,
     'Block or cascade? Iceberg spec requires empty namespace')

# 8. Delete namespace WITH child namespace
r = requests.delete(f'{CAT}/ns-test-catalog/namespaces/parent%1Fchild', headers=h(SVC_TOKEN))
test('Namespace', 'Delete namespace with child namespace', r, 400,
     'Block or cascade?')

# 9. Delete leaf namespace (empty)
r = requests.delete(f'{CAT}/ns-test-catalog/namespaces/parent%1Fchild%1Fgrandchild',
    headers=h(SVC_TOKEN))
test('Namespace', 'Delete leaf namespace (empty)', r, 204)

# Cleanup
requests.delete(f'{CAT}/ns-test-catalog/namespaces/parent/tables/dep-test-table',
    headers=h(SVC_TOKEN))
requests.delete(f'{CAT}/ns-test-catalog/namespaces/parent%1Fchild', headers=h(SVC_TOKEN))
requests.delete(f'{CAT}/ns-test-catalog/namespaces/parent', headers=h(SVC_TOKEN))
requests.delete(f'{BASE}/principals/ns-test-svc', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/principal-roles/ns-test-pr', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/catalogs/ns-test-catalog', headers=h(TOKEN_ROOT))
````

## Group 7 — Table Dependencies

````python
print('=== Group 7: Table Dependencies ===')

# Setup
requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json={'catalog': {'name': 'tbl-test-catalog', 'type': 'INTERNAL',
        'properties': {'default-base-location': 's3a://data-catalog-bucket/tbl-test/'},
        'storageConfigInfo': {'storageType': 'S3',
            'allowedLocations': ['s3a://data-catalog-bucket/'],
            'roleArn': 'arn:aws:iam::123456789012:role/dummy',
            'pathStyleAccess': True}}})
requests.post(f'{BASE}/catalogs/tbl-test-catalog/catalog-roles', headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'tbl-admin-role'}})
requests.put(f'{BASE}/catalogs/tbl-test-catalog/catalog-roles/tbl-admin-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'catalog', 'privilege': 'CATALOG_MANAGE_CONTENT'}})
requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'tbl-test-pr'}})
requests.put(f'{BASE}/principal-roles/tbl-test-pr/catalog-roles/tbl-test-catalog',
    headers=h(TOKEN_ROOT), json={'catalogRole': {'name': 'tbl-admin-role'}})
r = requests.post(f'{BASE}/principals', headers=h(TOKEN_ROOT),
    json={'name': 'tbl-test-svc', 'type': 'SERVICE'})
creds = r.json().get('credentials', {})
requests.put(f'{BASE}/principals/tbl-test-svc/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'tbl-test-pr'}})
SVC_TOKEN2 = get_token(creds.get('clientId', ''), creds.get('clientSecret', '')).json().get('access_token', '')
requests.post(f'{CAT}/tbl-test-catalog/namespaces', headers=h(SVC_TOKEN2),
    json={'namespace': ['tbl-ns'], 'properties': {}})

def make_table(name):
    return {'name': name,
            'schema': {'type': 'struct', 'schema-id': 0,
                'fields': [{'id': 1, 'name': 'id', 'type': 'long', 'required': True},
                           {'id': 2, 'name': 'value', 'type': 'string', 'required': False}]},
            'partition-spec': {'spec-id': 0, 'fields': []},
            'write-order': {'order-id': 0, 'fields': []},
            'stage-create': False, 'properties': {}}

# 1. Create table
r = requests.post(f'{CAT}/tbl-test-catalog/namespaces/tbl-ns/tables',
    headers=h(SVC_TOKEN2), json=make_table('test-table'))
test('Table', 'Create table', r, 200)

# 2. Create duplicate table
r = requests.post(f'{CAT}/tbl-test-catalog/namespaces/tbl-ns/tables',
    headers=h(SVC_TOKEN2), json=make_table('test-table'))
test('Table', 'Create duplicate table', r, 409)

# 3. Create table in non-existent namespace
r = requests.post(f'{CAT}/tbl-test-catalog/namespaces/nonexistent-ns/tables',
    headers=h(SVC_TOKEN2), json=make_table('orphan-table'))
test('Table', 'Create table in non-existent namespace', r, 404)

# 4. Create view with invalid SQL (does Polaris validate?)
r = requests.post(f'{CAT}/tbl-test-catalog/namespaces/tbl-ns/views',
    headers=h(SVC_TOKEN2),
    json={'name': 'invalid-view',
          'schema': {'type': 'struct', 'schema-id': 0,
              'fields': [{'id': 1, 'name': 'id', 'type': 'long', 'required': True}]},
          'view-version': {'version-id': 1, 'schema-id': 0, 'timestamp-ms': 0,
              'summary': {'engine-name': 'spark'},
              'default-namespace': ['tbl-ns'],
              'representations': [{'type': 'sql',
                  'sql': 'SELECT * FROM completely_nonexistent_table_xyz',
                  'dialect': 'spark'}]},
          'properties': {}})
test('Table', 'Create view with invalid SQL (references non-existent table)', r, 200,
     'Polaris stores view definition only — does NOT validate SQL')

# 5. Add table-scoped grant then delete table
requests.post(f'{BASE}/catalogs/tbl-test-catalog/catalog-roles', headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'tbl-reader-role'}})
requests.put(f'{BASE}/catalogs/tbl-test-catalog/catalog-roles/tbl-reader-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'table', 'namespace': ['tbl-ns'],
                    'tableName': 'test-table', 'privilege': 'TABLE_READ_DATA'}})

r = requests.delete(f'{CAT}/tbl-test-catalog/namespaces/tbl-ns/tables/test-table',
    headers=h(SVC_TOKEN2))
test('Table', 'Delete table with table-scoped grant pointing to it', r, 204,
     'Does Polaris block delete or leave orphaned grant?')

# Check if grant still exists after table deletion
r = requests.get(
    f'{BASE}/catalogs/tbl-test-catalog/catalog-roles/tbl-reader-role/grants',
    headers=h(TOKEN_ROOT))
grants = r.json().get('grants', [])
print(f'  💡 Grants remaining after table deletion: {grants}')
print(f'     → Orphaned grants left in grant_records: {len(grants) > 0}')

# 6. Delete view
r = requests.delete(f'{CAT}/tbl-test-catalog/namespaces/tbl-ns/views/invalid-view',
    headers=h(SVC_TOKEN2))
test('Table', 'Delete view', r, 403)

# Cleanup
requests.delete(f'{CAT}/tbl-test-catalog/namespaces/tbl-ns', headers=h(SVC_TOKEN2))
requests.delete(f'{BASE}/principals/tbl-test-svc', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/principal-roles/tbl-test-pr', headers=h(TOKEN_ROOT))
requests.delete(f'{BASE}/catalogs/tbl-test-catalog', headers=h(TOKEN_ROOT))
````

## Group 8 — Full Lifecycle Sequential Flow

````python
print('=== Group 8: Full Lifecycle Sequential Flow ===')
print('Complete create → use → delete in correct dependency order\n')

# ── CREATE PHASE ────────────────────────────────────────────
print('── CREATE PHASE ──')

r = requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'lifecycle-pr'}})
test('Lifecycle', 'Create principal role', r, 201)

r = requests.post(f'{BASE}/principals', headers=h(TOKEN_ROOT),
    json={'name': 'lifecycle-svc', 'type': 'SERVICE'})
test('Lifecycle', 'Create principal', r, 201)
lc_creds = r.json().get('credentials', {})

r = requests.put(f'{BASE}/principals/lifecycle-svc/principal-roles', headers=h(TOKEN_ROOT),
    json={'principalRole': {'name': 'lifecycle-pr'}})
test('Lifecycle', 'Assign principal role', r, 201)

r = requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
    json={'catalog': {'name': 'lifecycle-catalog', 'type': 'INTERNAL',
        'properties': {'default-base-location': 's3a://data-catalog-bucket/lifecycle/'},
        'storageConfigInfo': {'storageType': 'S3',
            'allowedLocations': ['s3a://data-catalog-bucket/'],
            'roleArn': 'arn:aws:iam::123456789012:role/dummy',
            'pathStyleAccess': True}}})
test('Lifecycle', 'Create catalog', r, 201)

r = requests.post(f'{BASE}/catalogs/lifecycle-catalog/catalog-roles', headers=h(TOKEN_ROOT),
    json={'catalogRole': {'name': 'lifecycle-role'}})
test('Lifecycle', 'Create catalog role', r, 201)

r = requests.put(f'{BASE}/catalogs/lifecycle-catalog/catalog-roles/lifecycle-role/grants',
    headers=h(TOKEN_ROOT),
    json={'grant': {'type': 'catalog', 'privilege': 'CATALOG_MANAGE_CONTENT'}})
test('Lifecycle', 'Grant privilege', r, 201)

r = requests.put(f'{BASE}/principal-roles/lifecycle-pr/catalog-roles/lifecycle-catalog',
    headers=h(TOKEN_ROOT), json={'catalogRole': {'name': 'lifecycle-role'}})
test('Lifecycle', 'Assign catalog role to principal role', r, 201)

# Get service token
LC_TOKEN = get_token(lc_creds.get('clientId', ''), lc_creds.get('clientSecret', '')).json().get('access_token', '')
print(f'  ✅ Service token obtained')

r = requests.post(f'{CAT}/lifecycle-catalog/namespaces', headers=h(LC_TOKEN),
    json={'namespace': ['lc-ns'], 'properties': {}})
test('Lifecycle', 'Create namespace', r, 200)

r = requests.post(f'{CAT}/lifecycle-catalog/namespaces', headers=h(LC_TOKEN),
    json={'namespace': ['lc-ns', 'nested'], 'properties': {}})
test('Lifecycle', 'Create nested namespace', r, 200)

r = requests.post(f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested/tables',
    headers=h(LC_TOKEN),
    json={'name': 'lifecycle-table',
          'schema': {'type': 'struct', 'schema-id': 0,
              'fields': [{'id': 1, 'name': 'id', 'type': 'long', 'required': True}]},
          'partition-spec': {'spec-id': 0, 'fields': []},
          'write-order': {'order-id': 0, 'fields': []},
          'stage-create': False, 'properties': {}})
test('Lifecycle', 'Create table in nested namespace', r, 200)

r = requests.post(f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested/views',
    headers=h(LC_TOKEN),
    json={'name': 'lifecycle-view',
          'schema': {'type': 'struct', 'schema-id': 0,
              'fields': [{'id': 1, 'name': 'id', 'type': 'long', 'required': True}]},
          'view-version': {'version-id': 1, 'schema-id': 0, 'timestamp-ms': 0,
              'summary': {'engine-name': 'spark'},
              'default-namespace': ['lc-ns', 'nested'],
              'representations': [{'type': 'sql',
                  'sql': 'SELECT id FROM lifecycle_table',
                  'dialect': 'spark'}]},
          'properties': {}})
test('Lifecycle', 'Create view', r, 200)

# ── USE PHASE ───────────────────────────────────────────────
print('\n── USE PHASE ──')

r = requests.get(f'{CAT}/lifecycle-catalog/namespaces', headers=h(LC_TOKEN))
test('Lifecycle', 'List namespaces', r, 200)

r = requests.get(f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested/tables',
    headers=h(LC_TOKEN))
test('Lifecycle', 'List tables', r, 200)

r = requests.get(
    f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested/tables/lifecycle-table',
    headers=h(LC_TOKEN))
test('Lifecycle', 'Load table metadata', r, 200)

# ── DELETE PHASE (reverse order) ────────────────────────────
print('\n── DELETE PHASE (reverse order) ──')

r = requests.delete(
    f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested/views/lifecycle-view',
    headers=h(LC_TOKEN))
test('Lifecycle', 'Delete view', r, 204)

r = requests.delete(
    f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested/tables/lifecycle-table',
    headers=h(LC_TOKEN))
test('Lifecycle', 'Delete table', r, 204)

r = requests.delete(f'{CAT}/lifecycle-catalog/namespaces/lc-ns%1Fnested',
    headers=h(LC_TOKEN))
test('Lifecycle', 'Delete nested namespace (now empty)', r, 204)

r = requests.delete(f'{CAT}/lifecycle-catalog/namespaces/lc-ns', headers=h(LC_TOKEN))
test('Lifecycle', 'Delete parent namespace (now empty)', r, 204)

r = requests.delete(
    f'{BASE}/catalogs/lifecycle-catalog/catalog-roles/lifecycle-role',
    headers=h(TOKEN_ROOT))
test('Lifecycle', 'Delete catalog role', r, 200)

r = requests.delete(f'{BASE}/catalogs/lifecycle-catalog', headers=h(TOKEN_ROOT))
test('Lifecycle', 'Delete catalog', r, 200)

r = requests.delete(f'{BASE}/principals/lifecycle-svc', headers=h(TOKEN_ROOT))
test('Lifecycle', 'Delete principal', r, 200)

r = requests.delete(f'{BASE}/principal-roles/lifecycle-pr', headers=h(TOKEN_ROOT))
test('Lifecycle', 'Delete principal role', r, 200)

print('\n✅ Full lifecycle complete')
````

## Group 9 — Error Response Format

````python
print('=== Group 9: Error Response Format ===')
print('Verify consistent error structure across all HTTP status codes\n')

error_cases = [
    ('400 Bad Request',   requests.post(f'{BASE}/catalogs', headers=h(TOKEN_ROOT),
                          json={'catalog': {'name': '', 'type': 'INTERNAL'}})),
    ('401 Unauthorized',  requests.get(f'{BASE}/principals',
                          headers={'Authorization': 'Bearer INVALID', 'Polaris-Realm': REALM})),
    ('403 Forbidden',     requests.get(f'{CAT}/test-catalog/namespaces', headers=h(TOKEN_ROOT))),
    ('404 Not Found',     requests.get(f'{BASE}/catalogs/nonexistent', headers=h(TOKEN_ROOT))),
    ('409 Conflict',      requests.post(f'{BASE}/principal-roles', headers=h(TOKEN_ROOT),
                          json={'principalRole': {'name': 'service_admin'}})),
]

print(f'{"Status":<20} {"Type":<40} {"Message":<60}')
print('-' * 120)
for label, r in error_cases:
    try:
        body = r.json()
        err  = body.get('error', body)
        etype   = err.get('type', 'N/A')[:40]
        emsg    = err.get('message', str(body))[:60]
        ecode   = err.get('code', r.status_code)
        print(f'{label:<20} {etype:<40} {emsg:<60}')
        print(f'{"":<20} HTTP={r.status_code}, error.code={ecode}')
    except:
        print(f'{label:<20} Non-JSON response: {r.text[:80]}')
    print()

print('\n💡 Error response structure:')
print(json.dumps({'error': {'message': '...', 'type': 'ExceptionClass', 'code': 404}}, indent=2))
````
